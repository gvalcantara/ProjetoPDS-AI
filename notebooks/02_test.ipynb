```python
MODELO = "/home/gustavo/_dev/GIT/ProjetoPDS-AI/weights/best.pt"
VIDEO = "/home/gustavo/_dev/GIT/ProjetoPDS-AI/SoccerTrack v2/mot/clips/117093.mp4"
```

```python
from ultralytics import YOLO
from pathlib import Path
from datetime import datetime
import torch

assert Path(MODELO).is_file(), "Confira o caminho do modelo."
assert Path(VIDEO).is_file(), "Confira o caminho do vídeo."

model = YOLO(MODELO)

# Identifica o teste pelo modelo, vídeo e horário
nome = (
    f"{Path(MODELO).parent.parent.name}_"
    f"{Path(VIDEO).stem}_"
    f"{datetime.now():%Y%m%d_%H%M%S}"
)

for resultado in model.predict(
    source=VIDEO,
    imgsz=640,                             # Resolução de entrada da rede
    conf=0.25,                             # Limiar mínimo de confiança
    device=0 if torch.cuda.is_available() else "cpu",
    save=True,                             # Salva o vídeo com as detecções
    stream=True,                           # Processa sem acumular todos os resultados na RAM
    project="/content/drive/MyDrive/UFSC/Projeto Nível I em Cont e Proc de Sinais I/yolo_testes",
    name=nome,
    verbose=False,
):
    pass

PASTA_RESULTADO = Path(model.predictor.save_dir)
print("Resultado salvo em:", PASTA_RESULTADO)
```
