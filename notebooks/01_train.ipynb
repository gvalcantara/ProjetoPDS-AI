```python
from ultralytics import YOLO
import cv2
import os
from google.colab import drive
from collections import defaultdict
drive.mount('/content/drive')
```

```python
DATASET = "/content/drive/MyDrive/UFSC/Projeto Nível I em Cont e Proc de Sinais I/dataset"
DATA_YAML = f"{DATASET}/dataset.yaml"
```

```python
model = YOLO("yolo26n.pt")

model.train(
    data=DATA_YAML,  # YAML com caminhos dos dados e classes
    device=0,              # Usa a primeira GPU; exige GPU disponível
    imgsz=640,             # Tamanho-alvo da entrada: 640 × 640, preservando proporções
    batch=-1,              # Ajusta automaticamente o lote à memória da GPU
    epochs=30,             # Limite de 30 passagens pelo conjunto de treino
    patience=8,            # Para após 8 épocas sem melhora na métrica de validação
    amp=True,              # Usa precisão mista para reduzir memória e acelerar na GPU
    workers=2,             # Usa 2 processos auxiliares para carregar e preparar imagens
    cache=False,           # Não armazena imagens em cache adicional na RAM ou no disco
    optimizer="auto",      # Escolhe automaticamente o otimizador e a taxa de aprendizado
    close_mosaic=5,         # Desativa mosaic nas últimas 5 épocas previstas
    seed=42,               # Define a semente aleatória; ajuda a reproduzir o experimento
    save=True,             # Salva checkpoints e os pesos do modelo
    project="/content/drive/MyDrive/yolo_runs",  # Pasta persistente dos resultados
    name="players_n_640_baseline",              # Nome da subpasta deste experimento
)
```
